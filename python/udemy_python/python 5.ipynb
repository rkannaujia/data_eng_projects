#### 1. Python Program to Find Largest Number in a List

```python
l=[3,4,44,5,66,7]
# inbuilt method 

max(l)
```

```python
# withot inbuilt method 
l=[3,4,44,5,66,7]

largest = 0
for i in l:
    if i>largest:
        largest =i 
print(largest)
```

 
#### 2. Python Program to Find LOWEST Number in a List

# List Comprehension

#### List comprehension offers a shorter syntax when you want to create a new list based on the values of an existing list.

#### This helps us to write cleaner, more readable code compared to traditional looping techniques.

### Advantages of List Comprehension:

#### Concise: Reduces multiple lines of code to a single line.
#### Readable: Easier to understand compared to traditional loops for simple operations.
#### Efficient: Faster than equivalent for loops due to optimized implementation.

```python
## Create a List of Squares:
a=[1,3,4,5,6,7]
l=[]
for i in a:
    l.append(i*i)
l
```

```python
[i*i for i in a]
```

```python
## Filter Even Numbers:
a=[1,3,4,5,6,7]
l=[]
for i in a:
    if i%2==0:
        l.append(i)
l
```

```python
[i for i in a if i%2==0]
```

```python
a=['hello','world']
l=[]
for i in a:
    l.append(i.upper())
l
```

```python
[i.upper() for i in a]
```
