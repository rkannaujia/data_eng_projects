## OOPS 

#### OOP stands for Object-Oriented Programming.
#### It’s a way of writing code that focuses on objects and classes instead of just actions (like in procedural programming).

#### Think of it as organizing your code around real-world things or concepts.

## OOP vs Procedural Programming

### Procedural Programming:

#### Focuses on steps or procedures to solve a problem.

#### Code is organized as a list of instructions.

#### Example: A recipe for making tea. You follow steps one after another (boil water, add tea leaves, add sugar, etc.).

### OOP:

#### Focuses on objects that represent real-world things.

#### Code is organized around classes and objects.

#### Example: Think of a car. A car has properties (color, model, speed) and actions (drive, brake, honk). In OOP, you create a "car" object with these properties and actions.

## Class and Object

### Class (Blueprint): 
#### A blueprint or template for creating objects.
#### It defines the properties (attributes) and actions (methods) that objects will have.

#### Example:Think of a class as a "car design" in a factory

#### A car blueprint defines:

##### Properties: What a car has (e.g., color, model, speed).

##### Actions: What a car can do (e.g., drive, brake, honk).

##### It’s just a plan or template; it doesn’t exist physically.

### Object (Real Car):

##### An object is the actual car built from the blueprint.

##### It has specific values for the properties (e.g., a red Toyota with a speed of 60 km/h).

##### It can perform actions (e.g., drive, brake).



```python
class Car:
    def __init__(self, color, model):  # Constructor to initialize properties
        self.color = color  # Property 1
        self.model = model  # Property 2
        self.speed = 0      # Property 3

    def drive(self):        # Method 1
        self.speed = 60
        print(f"The {self.color} {self.model} is driving at {self.speed} km/h.")

    def brake(self):        # Method 2
        self.speed = 0
        print(f"The {self.color} {self.model} has stopped.")
```

```python
# Create objects from the Car class
car1 = Car("Red", "Toyota")  # A red Toyota
car2 = Car("Blue", "Honda")  # A blue Honda

# Use the objects
car1.drive()  # Output: The Red Toyota is driving at 60 km/h.
car2.brake()  # Output: The Blue Honda has stopped.
```

## Why OOP?
#### It’s easier to manage and organize code for large projects.

#### You can reuse code (e.g., create multiple cars from the same class).

#### It models real-world problems more naturally.
