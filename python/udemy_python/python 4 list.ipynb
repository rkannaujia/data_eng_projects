# 9 . List in Python

#### A list is a built-in data structure in Python used to store multiple items in a single variable. 
#### Lists are ordered, mutable (modifiable), and can contain elements of different data types.


### Key Features of Lists
#### Ordered: Elements have a defined order and can be accessed by their index.
#### Mutable: Elements can be changed, added, or removed.
#### Heterogeneous: Can store elements of different types (integers, strings, floats, etc.).
#### Dynamic Size: Lists can grow or shrink as elements are added or removed.

```python
l =['a',3,4,'apple']
l.append(6)
l
```

```python
# create list 

a = [1,3,4,5,'apple']
type(a)
```

```python
# create empty list 
a = list()
type(a)
a
```

```python
# ordered - index 
a= [1,3,4,5,6,7,9]
# index start - 0 
# index end =-1 
a[-2]
```

```python
# mutable
a= [1,3,4,5,6,7,9]
a[0]=7
a
```

```python
# heterogeneous
a=[1,'app',4.0]
type(a)
```

```python
# dynamic size 
a.append(5)
a
```

###  list methods 

```python
# append - Adds an element to the end of the list

a= [1,3,4,5,6]
a.append(9)
a
```

```python
# insert - Inserts an element at a specific index
a.insert(3,10)
a
```

```python
# remove - Removes the first occurrence of a value
a.remove(10)
a
```

```python
# pop()	Removes an element by index and returns it
a
```

```python
a.pop()
```

```python
# sort()	Sorts the list in ascending order
l = [4,5,3,7,9]
l.sort()
l
```

```python
# reverse()	Reverses the list
l.reverse()
l
```

```python
# extend()	Adds elements from another list
l=[1,3,4]
l1=[4,5,6]
l.extend(l1)
l
```

```python
# index()	Returns the index of the first occurrence of a value

l.index(6)
```

```python
# count()	Counts occurrences of a value
l
```

```python
l.count(4)
```
