
# Lambda Functions in Python 🚀

#### A lambda function in Python is a small, anonymous function that can have multiple arguments but only a single expression. 
#### It is often used for short, simple operations

### lambda arguments: expression

#### lambda is the keyword.
#### Arguments are the input values.
#### Expression is the single operation to be performed.


```python
# write a fnction take 2 variable and sum it 
def summation(a,b):
    final=a+b
    return final

summation(5,7)
```

```python
x = lambda a,b : a+b
x(5,7)
```

```python
# write a fnction which will mltiply 3 varibales as inpt 
def multiply(a,b,c):
    mul=a*b*c
    return mul

multiply(2,3,4)
```

```python
x= lambda a,b,c: a*b*c
x(2,3,4)
```
