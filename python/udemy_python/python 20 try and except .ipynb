#  try-except

#### Python provides try and except blocks to handle error situations. In case an error occurs in try-block, 
#### Python stops executing try block adn jumps to exception block.
#### These blocks let you handle the errors without crashing the program.

```python
def divide(x,y):
    result = x//y
    return result

divide(18,0)
```

```python
def divide(x,y):
    try:
        result = x//y
        return result
    except:
        print('error occuring')
divide(18,0)
```

```python
def divide(x,y):
    try:
        result = x//y
        return result
    except Exception as e:
        print('error occuring--',e)
divide(18,0)
```

```python
def divide(x,y):
    try:
        result = x//y
        return result
    except ZeroDivisionError as e:
        print('error occuring--',e)
divide(18,0)
```
