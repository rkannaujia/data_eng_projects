# Tuple

#### Tuples are used to store multiple items in a single variable.

#### Tuple is one of 4 built-in data types in Python used to store collections of data, the other 3 are List, Set, and Dictionary, all with different qualities and usage.

#### A tuple is a collection which is ordered and unchangeable.

#### Tuples are written with round brackets.

## key feature

#### Tuple items are ordered, unchangeable, and allow duplicate values.

#### Tuple items are indexed, the first item has index [0], the second item has index [1] etc.

### Ordered
#### When we say that tuples are ordered, it means that the items have a defined order, and that order will not change.

### Unchangeable
#### Tuples are unchangeable, meaning that we cannot change, add or remove items after the tuple has been created.

### Allow Duplicates
#### Since tuples are indexed, they can have items with the same value:

## Why Use Tuples?

#### Immutability: Ensures that data remains unchanged, ideal for constants.
#### Faster Access: Tuples are faster than lists due to their immutability.

#### Tuples are a valuable tool for grouping related data while maintaining integrity and improving performance.

```python
a = ("apple", "banana", "cherry", "apple", "cherry")
a[0]='banana'
```

```python
# create tuple
a= (1,'abc',55.0)
type(a)
```

```python
# access tuple - indexing 
# starts - 0 
# ends - -1
a= (1,'abc',55.0)
a[-1]
```

```python
# unchangeable, or immutable 
a= (1,'abc',55.0)
a.append(5)
```

```python
# Since tuples are immutable, they do not have a built-in append() method, 
# but there are other ways to add items to a tuple.
# 1. Convert into a list: 
# 2. Add tuple to a tuple
a= (1,'abc',55.0)
a = list(a)
a.append(55)
a=tuple(a)

b = (1,3,4)
c=(1,)
b=b+c
b
```

```python
# we cant remove from tuple
b.remove(4)
```

```python
# del -we can delete the tuple
del b
```
