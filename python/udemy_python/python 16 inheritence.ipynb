# Inheritance

## Inheritance is a key Object-Oriented Programming (OOP) concept in Python 
##  that allows one class (child class) to reuse the properties and methods of another class (parent class)

###  Helps in code reusability and makes large projects more manageable.

```python
class A:
    def val(self):
        print('val method')

class B(A):
    def val1(self):
        print('val 1 method')

obj=B()
obj.val1()
obj.val()
```

```python
# mltilevel inheritence 

class A:
    def val(self):
        print('val method')
class B(A):
    def val4(self):
        print('val method')

class C(B):
    def val3(self):
        print('val3 method')

obj =C()
obj.val3()
obj.val4()
obj.val()
```

```python
# mltiple inheritence
```

```python

class A:
    def val(self):
        print('val method')
class B:
    def val4(self):
        print('val method')

class C(A,B):
    def val3(self):
        print('val3 method')

obj =C()
obj.val3()
obj.val4()
obj.val()
```
