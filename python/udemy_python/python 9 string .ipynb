# Strings
#### Strings in python are surrounded by either single quotation marks, or double quotation marks.

## Key Features of Strings in Python

#### Strings are immutable (cannot be changed after creation).
#### Powerful built-in methods for text manipulation.
#### Widely used in file handling, data processing, and user interfaces.

```python
print("Hello's")
```

```python
print('Hello')
```

```python
print('''hello ''')
```

```python
print(''' Lorem ipsum dolor sit amet,
consectetur adipiscing elit,
sed do eiusmod tempor incididunt
ut labore et dolore magna aliqua ''' )
```

```python
# upper 
a= 'hello world'
a.upper()
```

```python
#lower
b= 'HELLO WORLD'
b.lower()
```

```python
# remove whitespace
a= '   hello world   '
a.strip()
```

```python
# replace
a= 'hello world'
a.replace('world','hello')
```

```python
# split - method returns a list where the text between the specified separator becomes the list items.
a= 'hello world , how are you'
a.split()
```

```python
# String Concatenation
# To concatenate, or combine, two strings you can use the + operator.
a = 'hello'
b='manish'

a+' '+b
```

```python
# f string 
# we can combine strings and numbers by using f-strings or the format() method!

name = 'manish'

print(f"hello {name}")
```

```python
name = input("what is ur name")
print(f"hello {name}")
```
