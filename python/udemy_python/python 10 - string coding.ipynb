# question -1
#### Python Program to  Check if a String is Palindrome or Not
#### find reverse of the string and then Check if reverse and original are same or not.

#### Input : malayalam
#### Output : Yes



```python
x= 'malayalam'
rev= x[::-1]
rev
if x==rev:
    print('palindrome')
else:
    print('not')
```

```python
x= 'malayalam'
a=''
for i in x:
    a=i+a
if x==a:
    print('palindrome')
else:
    print('not')
```

# question -3
#### Python Program to  Words Frequency in String 

#### Input : test_str = 'hello is best' 
#### Output : {'hello': 1, 'is': 1, 'best': 1}





```python
s= 'hello is best is'
l=s.split()
l
d={}
for i in l:
    if i in d:
        d[i]+=1
    else:
        d[i]=1
d

```
