```python
## IF STATEMENTS 
## take user input and check IF uSER INPUT MARKS GREATER THAN 33 THEN PRINT YOU ARE PASS

num = int(input("enter your marks"))
if num>33:
   print('you are pass')

    
```

```python
## IF - else  STATEMENTS 

num = int(input("enter your marks"))
if num>33:
   print('you are pass')
else:
    print('failed on the exam')
    
```

```python
# if elif else 
## if num >80 grade a num >60 thne garade b num >40 grade c else fail

num = int(input("enter your marks"))
if num>80:
    print('grade A')
elif num>60:
    print('grade B')
elif num>40:
    print('grade C')
else:
    print('fail')



```

```python
# nested if 
# num = 0 then print 0 num<0 print negative num>0 then even or odd check 

num = int(input("enter your marks"))
if num>0:
    print('positive number')
    if num%2==0:
        print('even')
    else:
        print('odd')
elif num==0:
    print('zero number')
else:
    print('negative')




    
```
