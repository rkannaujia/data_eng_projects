# Date and DateTime in Python

#### Python provides the datetime module to work with dates and times. It allows us to manipulate and format date and time values efficiently.

```python
# Get today's date

from datetime import date

# Get today's date
today = date.today()
print("Today's Date:", today)

```

```python
# Create a specific date

custom_date = date(2024, 2, 6)
print("Custom Date:", custom_date)
```

```python
# Creating a DateTime Object

from datetime import datetime

# Get the current date and time
now = datetime.now()
print("Current Date & Time:", now)

```

```python
# Create a specific datetime
custom_datetime = datetime(2024, 2, 6, 14, 30, 15)
print("Custom DateTime:", custom_datetime)
```

```python
#  Formatting Date and Time
formatted_date = now.strftime("%Y-%m-%d %H:%M:%S")
print("Formatted DateTime:", formatted_date)
```

```python
# Converting String to DateTime
date_string = "2024-02-06 14:30:15"
print(type(date_string))
parsed_datetime = datetime.strptime(date_string, "%Y-%m-%d %H:%M:%S")
print("Parsed DateTime:", parsed_datetime)
print(type(parsed_datetime))
```

```python
# Performing Date Arithmetic

from datetime import timedelta

# Add 5 days
future_date = today + timedelta(days=5)
print("Future Date:", future_date)

# Subtract 2 hours
past_time = now - timedelta(hours=2)
print("Past Time:", past_time)
```
