# *args and **kwargs in Python

#### In Python, *args and **kwargs allow you to pass a variable number of arguments to a function.



```python
def adder(a,b,c):
    return a+b+c
```

```python
adder(3,4)
```

# 1️⃣ *args (Non-Keyword Arguments)

#### If you do not know how many arguments that will be passed into your function, add a * before the parameter name in the function definition.

#### This way the function will receive a tuple of arguments, and can access the items accordingly:

```python
def adder(*num):
    a=0
    for i in num:
        a+=i
    return a
    
```

```python
adder(3,4,5,6,7)
```

# 2️⃣ **kwargs (Keyword Arguments)

#### **kwargs allows a function to accept multiple keyword arguments as a dictionary.

```python
def naming(**kwargs):
    for key,val in kwargs.items():
        print(key,val)
    
```

```python
naming(name='manish',srname='tiwari',age=28)
```
