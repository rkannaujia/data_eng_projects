# Encapsulation

### it is one of the fundamental principles of Object-Oriented Programming (OOP) in Python. 
### It is the process of restricting direct access to certain details of an object and only exposing necessary functionalities.

#### we can restric direct access to method and variable 
#### in python we can access variable and method pblicly 

```python
# private variable

class car :
    __balance=10 
    def val(self):
        print(self.__balance)
        print('hello')

obj = car()
obj.val()
```

```python
class car:
    def __init__(self):
        self.__val()
    def __val(self):
        print('hello')

obj = car()
```

## two levels of restricted access:

#### Protected Members (_variable) → Can be accessed inside the class and subclasses.
#### Private Members (__variable) → Can only be accessed inside the class (name-mangled).

```python
__a - private 
_a - protected 

```
