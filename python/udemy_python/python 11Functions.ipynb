#  Functions

#### A function in Python is a reusable block of code that performs a specific task..
#### Functions help in modularizing code, improving readability, and avoiding repetition.

#### You can pass data, known as parameters, into a function.
#### A function can return data as a result.

## Key Benefits of Functions

#### ✔ Code reusability
#### ✔ Improves readability and modularity
#### ✔ Allows code organization
#### ✔ Supports default values and flexibility

```python
# create own function which will do sum of all the list elements

def summation(l):
    a=0
    for i in l:
        a+=i
    return a
l=[3,4,5,6,7]
summation(l)
```

```python
# create own function which will find max of all the list elements

def maximum(l):
    a=0
    for i in l:
        if i>a:
            a=i
    return a
l=[3,444,55,6,7]
maximum(l)
```
